```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python

x = np.linspace(0,5,11)
x 
y = x**3
y
```

```python
#basics of matplotlib

plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

```python
# Subplots in Matplotlib
plt.subplot(2,2,1)
plt.subplot(2,2,4)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,2)
```

```python
#Object Oriented Way 
fig = plt.figure(figsize= (20,8), dpi= 300)
axis1 = fig.add_axes([0.1,0.1,.4,.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.3])
axis2.plot(x,y)
```
